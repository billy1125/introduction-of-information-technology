# 一、進位轉換練習題（共 25 題）
## （一）十進位 → 二進位（5 題）

**方法**：除以 2，記錄餘數，直到商為 0，餘數由下往上排列即為二進位結果。

### 第 1 題：十進位 **58** 轉二進位

$$
\begin{aligned}
58 \div 2 &= 29 \ \text{餘}\ 0\\
29 \div 2 &= 14 \ \text{餘}\ 1\\
14 \div 2 &= 7 \ \text{餘}\ 0\\
7 \div 2 &= 3 \ \text{餘}\ 1\\
3 \div 2 &= 1 \ \text{餘}\ 1\\
1 \div 2 &= 0 \ \text{餘}\ 1
\end{aligned}
$$

由下往上排列餘數：$58_{10} = 111010_2$

### 第 2 題：十進位 **75** 轉二進位

$$
\begin{aligned}
75 \div 2 &= 37 \ \text{餘}\ 1\\
37 \div 2 &= 18 \ \text{餘}\ 1\\
18 \div 2 &= 9 \ \text{餘}\ 0\\
9 \div 2 &= 4 \ \text{餘}\ 1\\
4 \div 2 &= 2 \ \text{餘}\ 0\\
2 \div 2 &= 1 \ \text{餘}\ 0\\
1 \div 2 &= 0 \ \text{餘}\ 1
\end{aligned}
$$

由下往上排列餘數：$75_{10} = 1001011_2$

### 第 3 題：十進位 **100** 轉二進位

$$
\begin{aligned}
100 \div 2 &= 50 \ \text{餘}\ 0\\
50 \div 2 &= 25 \ \text{餘}\ 0\\
25 \div 2 &= 12 \ \text{餘}\ 1\\
12 \div 2 &= 6 \ \text{餘}\ 0\\
6 \div 2 &= 3 \ \text{餘}\ 0\\
3 \div 2 &= 1 \ \text{餘}\ 1\\
1 \div 2 &= 0 \ \text{餘}\ 1
\end{aligned}
$$

由下往上排列餘數：$100_{10} = 1100100_2$

### 第 4 題：十進位 **37** 轉二進位

$$
\begin{aligned}
37 \div 2 &= 18 \ \text{餘}\ 1\\
18 \div 2 &= 9 \ \text{餘}\ 0\\
9 \div 2 &= 4 \ \text{餘}\ 1\\
4 \div 2 &= 2 \ \text{餘}\ 0\\
2 \div 2 &= 1 \ \text{餘}\ 0\\
1 \div 2 &= 0 \ \text{餘}\ 1
\end{aligned}
$$

由下往上排列餘數：$37_{10} = 100101_2$

### 第 5 題：十進位 **214** 轉二進位

$$
\begin{aligned}
214 \div 2 &= 107 \ \text{餘}\ 0\\
107 \div 2 &= 53 \ \text{餘}\ 1\\
53 \div 2 &= 26 \ \text{餘}\ 1\\
26 \div 2 &= 13 \ \text{餘}\ 0\\
13 \div 2 &= 6 \ \text{餘}\ 1\\
6 \div 2 &= 3 \ \text{餘}\ 0\\
3 \div 2 &= 1 \ \text{餘}\ 1\\
1 \div 2 &= 0 \ \text{餘}\ 1
\end{aligned}
$$

由下往上排列餘數：$214_{10} = 11010110_2$

## （二）二進位 → 十進位（5 題）

**方法**：每個位元乘上對應的 $2$ 的次方後相加，次方由最右邊的 $2^0$ 開始，往左遞增。

### 第 6 題：二進位 **10110₂** 轉十進位

$$
10110_2 = 1\times2^4+0\times2^3+1\times2^2+1\times2^1+0\times2^0 = 16+0+4+2+0 = 22_{10}
$$

### 第 7 題：二進位 **11010₂** 轉十進位

$$
11010_2 = 1\times2^4+1\times2^3+0\times2^2+1\times2^1+0\times2^0 = 16+8+0+2+0 = 26_{10}
$$

### 第 8 題：二進位 **111011₂** 轉十進位

$$
111011_2 = 1\times2^5+1\times2^4+1\times2^3+0\times2^2+1\times2^1+1\times2^0 = 32+16+8+0+2+1 = 59_{10}
$$

### 第 9 題：二進位 **100001₂** 轉十進位

$$
100001_2 = 1\times2^5+0\times2^4+0\times2^3+0\times2^2+0\times2^1+1\times2^0 = 32+0+0+0+0+1 = 33_{10}
$$

### 第 10 題：二進位 **111111₂** 轉十進位

$$
111111_2 = 1\times2^5+1\times2^4+1\times2^3+1\times2^2+1\times2^1+1\times2^0 = 32+16+8+4+2+1 = 63_{10}
$$

## （三）十進位小數 → 二進位（4 題）

**方法**：小數部分持續乘以 2，取整數部分作為二進位位元，剩下的小數部分繼續乘 2，直到小數為 0（或達到所需精度）。

### 第 11 題：十進位小數 **0.75** 轉二進位

$$
\begin{aligned}
0.75 \times 2 &= 1.5 \ \rightarrow\ \text{取}\ 1，餘\ 0.5\\
0.5 \times 2 &= 1.0 \ \rightarrow\ \text{取}\ 1，餘\ 0
\end{aligned}
$$

由上往下排列取出的整數部分：$0.75_{10} = 0.11_2$

### 第 12 題：十進位小數 **0.125** 轉二進位

$$
\begin{aligned}
0.125 \times 2 &= 0.25 \ \rightarrow\ \text{取}\ 0，餘\ 0.25\\
0.25 \times 2 &= 0.5 \ \rightarrow\ \text{取}\ 0，餘\ 0.5\\
0.5 \times 2 &= 1.0 \ \rightarrow\ \text{取}\ 1，餘\ 0
\end{aligned}
$$

$0.125_{10} = 0.001_2$

### 第 13 題：十進位小數 **0.375** 轉二進位

$$
\begin{aligned}
0.375 \times 2 &= 0.75 \ \rightarrow\ \text{取}\ 0，餘\ 0.75\\
0.75 \times 2 &= 1.5 \ \rightarrow\ \text{取}\ 1，餘\ 0.5\\
0.5 \times 2 &= 1.0 \ \rightarrow\ \text{取}\ 1，餘\ 0
\end{aligned}
$$

$0.375_{10} = 0.011_2$

### 第 14 題：十進位小數 **0.8125** 轉二進位

$$
\begin{aligned}
0.8125 \times 2 &= 1.625 \ \rightarrow\ \text{取}\ 1，餘\ 0.625\\
0.625 \times 2 &= 1.25 \ \rightarrow\ \text{取}\ 1，餘\ 0.25\\
0.25 \times 2 &= 0.5 \ \rightarrow\ \text{取}\ 0，餘\ 0.5\\
0.5 \times 2 &= 1.0 \ \rightarrow\ \text{取}\ 1，餘\ 0
\end{aligned}
$$

$0.8125_{10} = 0.1101_2$

## （四）二進位 → 十六進位（4 題）

**方法**：從右往左每 4 個位元一組，不足 4 位在左邊補 0，再將每組轉換為對應的十六進位數字。

### 第 15 題：二進位 **01111001₂** 轉十六進位

$$
\underbrace{0111}_{7}\ \underbrace{1001}_{9} \ \rightarrow\ 01111001_2 = 0\text{x}79
$$

### 第 16 題：二進位 **11001010₂** 轉十六進位

$$
\underbrace{1100}_{C}\ \underbrace{1010}_{A} \ \rightarrow\ 11001010_2 = 0\text{x}CA
$$

### 第 17 題：二進位 **11110000₂** 轉十六進位

$$
\underbrace{1111}_{F}\ \underbrace{0000}_{0} \ \rightarrow\ 11110000_2 = 0\text{x}F0
$$

### 第 18 題：二進位 **01011100₂** 轉十六進位

$$
\underbrace{0101}_{5}\ \underbrace{1100}_{C} \ \rightarrow\ 01011100_2 = 0\text{x}5C
$$

## （五）十進位 → 十六進位（4 題）

**方法**：除以 16，記錄餘數（10~15 對應 A~F），直到商為 0，餘數由下往上排列即為十六進位結果。

### 第 19 題：十進位 **91** 轉十六進位

$$
\begin{aligned}
91 \div 16 &= 5 \ \text{餘}\ 11\ (\text{B})\\
5 \div 16 &= 0 \ \text{餘}\ 5
\end{aligned}
$$

由下往上排列：$91_{10} = 0\text{x}5B$

### 第 20 題：十進位 **171** 轉十六進位

$$
\begin{aligned}
171 \div 16 &= 10 \ \text{餘}\ 11\ (\text{B})\\
10 \div 16 &= 0 \ \text{餘}\ 10\ (\text{A})
\end{aligned}
$$

由下往上排列：$171_{10} = 0\text{x}AB$

### 第 21 題：十進位 **250** 轉十六進位

$$
\begin{aligned}
250 \div 16 &= 15 \ \text{餘}\ 10\ (\text{A})\\
15 \div 16 &= 0 \ \text{餘}\ 15\ (\text{F})
\end{aligned}
$$

由下往上排列：$250_{10} = 0\text{x}FA$

### 第 22 題：十進位 **45** 轉十六進位

$$
\begin{aligned}
45 \div 16 &= 2 \ \text{餘}\ 13\ (\text{D})\\
2 \div 16 &= 0 \ \text{餘}\ 2

$$

由下往上排列：$45_{10} = 0\text{x}2D$

## （六）十六進位 → 十進位（3 題）

**方法**：每個十六進位數字（A~F 對應 10~15）乘上對應的 $16$ 的次方後相加。

### 第 23 題：十六進位 **0x3D** 轉十進位

$$
0\text{x}3D = 3\times16^1+13\times16^0 = 48+13 = 61_{10}
$$

### 第 24 題：十六進位 **0x4E** 轉十進位

$$
0\text{x}4E = 4\times16^1+14\times16^0 = 64+14 = 78_{10}
$$

### 第 25 題：十六進位 **0x7B** 轉十進位

$$
0\text{x}7B = 7\times16^1+11\times16^0 = 112+11 = 123_{10}
$$

# 二、IEEE 754 浮點數練習題（共 5 題）

**公式回顧**：

$$
\text{數值} = (-1)^{\text{符號}} \times 1.\text{尾數} \times 2^{(\text{指數} - 127)}
$$

前 3 題為「十進位 → IEEE 754 單精度浮點數」，後 2 題為「IEEE 754 單精度浮點數 → 十進位」。

### 第 26 題：十進位 **8** 轉 IEEE 754 單精度浮點數

**Step 1**：正數，符號位 = $0$。

**Step 2**：$8_{10} = 1000_2 = 1.000_2 \times 2^3$。

**Step 3**：指數儲存值 $= 3+127=130=10000010_2$。

**Step 4**：尾數 $= 000$，補 0 至 23 位 $= 00000000000000000000000$。

結果：$0\ 10000010\ 00000000000000000000000$

### 第 27 題：十進位 **-3.25** 轉 IEEE 754 單精度浮點數

**Step 1**：負數，符號位 = $1$。

**Step 2**：$3.25_{10} = 11.01_2 = 1.101_2 \times 2^1$。

**Step 3**：指數儲存值 $= 1+127=128=10000000_2$。

**Step 4**：尾數 $= 101$，補 0 至 23 位 $= 10100000000000000000000$。

結果：$1\ 10000000\ 10100000000000000000000$

### 第 28 題：十進位 **0.5** 轉 IEEE 754 單精度浮點數

**Step 1**：正數，符號位 = $0$。

**Step 2**：$0.5_{10} = 1.0_2 \times 2^{-1}$。

**Step 3**：指數儲存值 $= -1+127=126=01111110_2$。

**Step 4**：尾數 $= 000$，補 0 至 23 位 $= 00000000000000000000000$。

結果：$0\ 01111110\ 00000000000000000000000$

### 第 29 題：IEEE 754 單精度浮點數 **0 10000001 01000000000000000000000** 轉十進位

**Step 1**：符號 $=0$（正數）。

**Step 2**：指數儲存值 $=10000001_2=129$，實際指數 $=129-127=2$。

**Step 3**：尾數 $01000000000000000000000 \rightarrow 0\times2^{-1}+1\times2^{-2}=0.25$，故有效數字 $=1.25$。

**Step 4**：數值 $=1.25\times2^2=5$。

結果：$5_{10}$

### 第 30 題：IEEE 754 單精度浮點數 **1 01111101 10000000000000000000000** 轉十進位

**Step 1**：符號 $=1$（負數）。

**Step 2**：指數儲存值 $=01111101_2=125$，實際指數 $=125-127=-2$。

**Step 3**：尾數 $10000000000000000000000 \rightarrow 1\times2^{-1}=0.5$，故有效數字 $=1.5$。

**Step 4**：數值 $=-(1.5\times2^{-2})=-0.375$。

結果：$-0.375_{10}$

# 三、Python 內建函數轉換範例

以上 30 題都是「手算」的計算過程。實務上 Python 標準函式庫已經內建了對應的轉換函數，不需要自己刻迴圈。以下針對**每一種轉換類型各挑一題**做示範，對照看看內建函數的結果是否與手算答案一致。

## 1. 十進位 → 二進位：`bin()`

以第 1 題（十進位 **58**）為例，`bin()` 會回傳以 `"0b"` 開頭的字串，用切片 `[2:]` 去掉開頭即可。

```python
數值 = 58
二進位結果 = bin(數值)[2:]
print(f"{數值} 轉二進位：{二進位結果}")
print(f"與手算答案 111010 是否相符：{二進位結果 == '111010'}")
```

## 2. 二進位 → 十進位：`int(x, 2)`

以第 6 題（二進位 **10110₂**）為例，`int()` 的第二個參數指定原始字串的進位基底。

```python
二進位字串 = "10110"
十進位結果 = int(二進位字串, 2)
print(f"{二進位字串}₂ 轉十進位：{十進位結果}")
print(f"與手算答案 22 是否相符：{十進位結果 == 22}")
```

## 3. 十進位小數 → 二進位：`fractions.Fraction`

以第 11 題（十進位小數 **0.75**）為例。Python 的浮點數本身就是以二進位儲存，`Fraction(小數)` 能把它還原成最簡分數；只要分母是 2 的次方，分子的二進位表示（補齊位數）就是小數點後的部分。

```python
from fractions import Fraction

小數值 = 0.75
分數 = Fraction(小數值)  # 例如 0.75 -> 3/4
位數 = 分數.denominator.bit_length() - 1  # 分母 = 2^位數
二進位小數 = format(分數.numerator, f"0{位數}b")
print(f"{小數值} = {分數.numerator}/{分數.denominator}")
print(f"{小數值} 轉二進位：0.{二進位小數}")
print(f"與手算答案 0.11 是否相符：{二進位小數 == '11'}")
```

## 4. 二進位 → 十六進位：`int()` + `hex()`

以第 15 題（二進位 **01111001₂**）為例，先用 `int(x, 2)` 轉成十進位整數，再用 `hex()` 轉成十六進位字串。

```python
二進位字串 = "01111001"
十六進位結果 = hex(int(二進位字串, 2))[2:].upper()
print(f"{二進位字串}₂ 轉十六進位：0x{十六進位結果}")
print(f"與手算答案 0x79 是否相符：{十六進位結果 == '79'}")
```

## 5. 十進位 → 十六進位：`hex()`

以第 19 題（十進位 **91**）為例。

```python
數值 = 91
十六進位結果 = hex(數值)[2:].upper()
print(f"{數值} 轉十六進位：0x{十六進位結果}")
print(f"與手算答案 0x5B 是否相符：{十六進位結果 == '5B'}")
```

## 6. 十六進位 → 十進位：`int(x, 16)`

以第 23 題（十六進位 **0x3D**）為例。

```python
十六進位字串 = "3D"
十進位結果 = int(十六進位字串, 16)
print(f"0x{十六進位字串} 轉十進位：{十進位結果}")
print(f"與手算答案 61 是否相符：{十進位結果 == 61}")
```

## 7. IEEE 754：十進位 → 位元，`struct` 模組

以第 26 題（十進位 **8**）為例。`struct.pack(">f", 數值)` 會依 IEEE 754 單精度規則，把浮點數打包成 4 個位元組（大端序），再用 `int.from_bytes()` 轉成整數並以 `format(..., "032b")` 顯示完整 32 個位元。

```python
import struct

數值 = 8.0
位元組 = struct.pack(">f", 數值)
整數值 = int.from_bytes(位元組, byteorder="big")
位元字串 = format(整數值, "032b")
符號, 指數, 尾數 = 位元字串[0], 位元字串[1:9], 位元字串[9:]
print(f"{數值} 的 IEEE 754 單精度表示：{符號} {指數} {尾數}")
print(f"與手算答案 0 10000010 00000000000000000000000 是否相符："
      f"{(符號, 指數, 尾數) == ('0', '10000010', '00000000000000000000000')}")
```

## 8. IEEE 754：位元 → 十進位，`struct` 模組

以第 29 題（位元 **0 10000001 01000000000000000000000**）為例。做法與上題相反：把 32 個位元組成的字串轉成整數、還原成 4 個位元組，再用 `struct.unpack(">f", ...)` 還原成浮點數。

```python
位元字串 = "0" + "10000001" + "01000000000000000000000"
整數值 = int(位元字串, 2)
位元組 = 整數值.to_bytes(4, byteorder="big")
數值 = struct.unpack(">f", 位元組)[0]
print(f"該 IEEE 754 位元表示的十進位數值：{數值}")
print(f"與手算答案 5.0 是否相符：{數值 == 5.0}")
```

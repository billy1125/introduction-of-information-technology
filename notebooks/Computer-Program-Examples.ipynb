## 第一章　程式設計基本概念

### 1-1　程式的本質：輸入 → 處理 → 輸出

```python
# 最簡單的程式結構示範：良率計算

# ── 輸入 ──
總產量 = 500
不良品數 = 12

# ── 處理 ──
良品數 = 總產量 - 不良品數
良率 = 良品數 / 總產量 * 100

# ── 輸出 ──
print(f"良品數：{良品數} 件")
print(f"良率：{良率:.1f}%")
```

### 1-3　基本運算

```python
# ── 算術運算 ──

總產量 = 500
不良品數 = 12

良品數 = 總產量 - 不良品數  # 減法
良率 = 良品數 / 總產量  # 除法
良率百分比 = 良率 * 100  # 乘法

# 取餘數（%）：常用於判斷整除、輪班計算
班次編號 = 7
所屬班別 = 班次編號 % 3  # 7 除以 3 餘 1 → 第1班

print(f"良品數：{良品數}")
print(f"良率：{良率:.4f}")
print(f"良率百分比：{良率百分比:.1f}%")
print(f"第 {班次編號} 班次屬於：第 {所屬班別} 班")
```

```python
# ── 邏輯運算 ──

溫度 = 23.5
壓力 = 102.0

溫度上限 = 25.0
溫度下限 = 20.0
壓力上限 = 105.0
壓力下限 = 98.0

# and：兩個條件都成立才為 True
溫度正常 = 溫度下限 <= 溫度 <= 溫度上限
壓力正常 = 壓力下限 <= 壓力 <= 壓力上限
製程正常 = 溫度正常 and 壓力正常

# or：任一條件成立即為 True
需要警報 = (not 溫度正常) or (not 壓力正常)

print(f"溫度 {溫度}°C 正常：{溫度正常}")
print(f"壓力 {壓力} kPa 正常：{壓力正常}")
print(f"製程正常：{製程正常}")
print(f"需要警報：{需要警報}")
```

```python
# 多條件組合：製程狀態綜合判斷

溫度 = 26.5  # ← 試著修改這個值
良率 = 0.93  # ← 試著修改這個值

if 溫度 > 30.0:
    print("🔴 緊急：溫度過高，立即停機檢查")
elif 溫度 > 25.0 and 良率 < 0.95:
    print("🟡 警告：溫度偏高且良率不達標，通知工程師")
elif 良率 < 0.95:
    print("🟡 警告：良率不達標，請確認製程參數")
else:
    print("🟢 正常：製程狀態良好")
```

### 2-2　for 迴圈

```python
# 用 for 迴圈處理一週的生產紀錄

星期 = ["週一", "週二", "週三", "週四", "週五"]
每日產量 = [480, 512, 498, 467, 521]
每日不良 = [12, 8, 15, 22, 9]

目標良率 = 95.0
總產量 = 0
低於目標天數 = 0

print(f"{'日期':4}  {'產量':>6}  {'不良':>4}  {'良率':>7}  {'狀態'}")
print("-" * 40)

for i in range(5):  # i 從 0 到 4
    良率 = (每日產量[i] - 每日不良[i]) / 每日產量[i] * 100
    總產量 += 每日產量[i]  # 累計總產量
    狀態 = "✓" if 良率 >= 目標良率 else "✗"
    if 良率 < 目標良率:
        低於目標天數 += 1
    print(f"{星期[i]:4}  {每日產量[i]:>6}  {每日不良[i]:>4}  {良率:>6.1f}%  {狀態}")

print("-" * 40)
print(f"週總產量：{總產量} 件")
print(f"低於目標天數：{低於目標天數} 天")
```

### 2-3　while 迴圈

```python
# 模擬生產線運作：生產到達配額為止

配額 = 100  # 今日生產目標
已生產 = 0
批次大小 = 12  # 每次加工一個批次
批次編號 = 0

print(f"今日配額：{配額} 件")
print("-" * 30)

while 已生產 < 配額:
    批次編號 += 1
    # 避免超產：本批次只生產到剛好達標
    本批產量 = min(批次大小, 配額 - 已生產)
    已生產 += 本批產量
    print(f"批次 {批次編號:02d}：生產 {本批產量} 件，累計 {已生產} 件")

print("-" * 30)
print(f"配額完成！共執行 {批次編號} 個批次")
```

### 2-4　break 與 continue

```python
# break：偵測到嚴重異常立即停止

溫度資料 = [22.1, 23.4, 21.8, None, 22.9, 31.5, 23.1]
#                                ↑            ↑
#                            感測器離線    緊急高溫

緊急上限 = 30.0

print("開始逐筆分析感測器資料...")

for i, 溫度 in enumerate(溫度資料):
    # continue：跳過空值（感測器離線）
    if 溫度 is None:
        print(f"  第 {i+1} 筆：感測器離線，略過")
        continue

    # break：溫度超過緊急上限，立即中止
    if 溫度 > 緊急上限:
        print(f"  第 {i+1} 筆：{溫度}°C ── 🔴 緊急高溫！停止分析，執行停機程序")
        break

    print(f"  第 {i+1} 筆：{溫度}°C ── 正常")

print("分析結束")
```

## 第三章　函數與模組化

### 3-1　定義與呼叫函數

```python
# 函數定義：計算良率
def 計算良率(總產量, 不良品數):
    """輸入總產量與不良品數，回傳良率百分比（0~100）"""
    if 總產量 == 0:
        return 0  # 避免除以零
    良品數 = 總產量 - 不良品數
    return 良品數 / 總產量 * 100


# 函數呼叫：同一個函數可以重複使用
A線良率 = 計算良率(500, 12)
B線良率 = 計算良率(320, 8)
C線良率 = 計算良率(410, 31)

print(f"A線良率：{A線良率:.1f}%")
print(f"B線良率：{B線良率:.1f}%")
print(f"C線良率：{C線良率:.1f}%")
```

```python
# 函數可以有預設參數值
def 判斷達標(良率, 目標=95.0):
    """判斷良率是否達到目標，目標預設為 95%"""
    if 良率 >= 目標:
        return True, "達標 ✓"
    else:
        差距 = 目標 - 良率
        return False, f"未達標 ✗（差 {差距:.1f}%）"


# 使用預設目標 95%
達標, 訊息 = 判斷達標(A線良率)
print(f"A線：{訊息}")

達標, 訊息 = 判斷達標(C線良率)
print(f"C線：{訊息}")

# 指定不同目標
達標, 訊息 = 判斷達標(C線良率, 目標=90.0)
print(f"C線（目標90%）：{訊息}")
```

### 3-2　模組化設計——將生產報表系統拆成多個函數

每個函數只負責一件事，最後在主程式中組合使用。

```python
# ── 模組1：計算單條生產線的統計數字 ──
def 分析生產線(名稱, 總產量, 不良品數, 目標良率=95.0):
    """回傳一條生產線的分析結果（字典格式）"""
    良率 = 計算良率(總產量, 不良品數)
    達標, 狀態 = 判斷達標(良率, 目標良率)
    return {"名稱": 名稱, "總產量": 總產量, "良率": 良率, "達標": 達標, "狀態": 狀態}


# ── 模組2：列印單條生產線的資訊 ──
def 列印生產線結果(結果):
    print(
        f"  {結果['名稱']}：產量 {結果['總產量']} 件，"
        f"良率 {結果['良率']:.1f}%  {結果['狀態']}"
    )


# ── 模組3：產生完整報表 ──
def 產生日報表(生產資料):
    """輸入多條生產線資料，輸出完整日報"""
    print("=" * 45)
    print("           生產日報表")
    print("=" * 45)

    所有良率 = []
    未達標線 = []

    for 線名, 產量, 不良 in 生產資料:
        結果 = 分析生產線(線名, 產量, 不良)
        列印生產線結果(結果)
        所有良率.append(結果["良率"])
        if not 結果["達標"]:
            未達標線.append(結果["名稱"])

    print("-" * 45)
    平均良率 = sum(所有良率) / len(所有良率)
    print(f"  全廠平均良率：{平均良率:.1f}%")
    if 未達標線:
        print(f"  ⚠ 需要改善：{未達標線}")
    else:
        print("  ✓ 所有生產線均達標")
    print("=" * 45)


# ── 主程式：呼叫報表函數 ──
今日資料 = [
    ("A線", 500, 12),
    ("B線", 320, 8),
    ("C線", 410, 31),
    ("D線", 280, 4),
]

產生日報表(今日資料)
```

### 4-1　順序儲存與索引

```python
# ── 列表就是一排排好順序的儲存格，每一格有一個索引 ──
儲位編號 = ["A-01", "A-02", "A-03", "A-04", "A-05"]

# 索引從 0 開始數，代表「第幾個」，不是儲位的名字
for 索引, 編號 in enumerate(儲位編號):
    print(f"索引 {索引} → 儲位 {編號}")

print(f"\n列表長度：{len(儲位編號)} 格")
print(f"第 1 格（索引 0）　：{儲位編號[0]}")
print(f"最後 1 格（索引 -1）：{儲位編號[-1]}")
```

### 4-2　列表（List）——儲存與操作一組資料

```python
# 建立列表
每日產量 = [480, 512, 498, 467, 521]

# 用索引存取特定元素（索引從 0 開始）
print(f"第1天（索引0）：{每日產量[0]}")
print(f"第3天（索引2）：{每日產量[2]}")
print(f"最後一天（索引-1）：{每日產量[-1]}")  # 負數索引從尾端算

# 常用的列表操作
print(f"\n列表長度：{len(每日產量)}")
print(f"最大值：{max(每日產量)}")
print(f"最小值：{min(每日產量)}")
print(f"總和：{sum(每日產量)}")
print(f"平均：{sum(每日產量)/len(每日產量):.1f}")
```

### 4-3　存取與走訪

**隨機存取**是「直接跳到第 n 格」，**走訪**是「從頭到尾跑一遍」。
兩者是列表最常用的操作，也是後面所有資料處理的基礎。

```python
每日產量 = [480, 512, 498, 467, 521, 489, 503]

# ── 隨機存取（Random Access）：直接跳到指定索引 ──
# 不管列表多長，取出任何一個元素的時間都一樣快
print("【隨機存取】")
print(f"第 3 天產量：{每日產量[2]} 件")
print(f"第 7 天產量：{每日產量[6]} 件")

# ── 走訪（Traversal）：從頭到尾，一個都不漏 ──
print("\n【走訪】")
累計 = 0
for 第幾天, 產量 in enumerate(每日產量, start=1):
    累計 += 產量
    print(f"第 {第幾天} 天：{產量:>4} 件，累計 {累計:>5} 件")
```

### 4-4　工業應用：每日產量紀錄

```python
# ── 工業應用：一個月 30 天的每日產量紀錄 ──
月產量紀錄 = [
    480, 512, 498, 467, 521, 489, 503, 476, 530, 495,
    508, 461, 517, 484, 499, 526, 470, 492, 511, 487,
    504, 468, 523, 496, 481, 515, 490, 502, 473, 519,
]

月總量 = sum(月產量紀錄)
月平均 = 月總量 / len(月產量紀錄)

# index() 會回傳該值第一次出現的索引，+1 才是「第幾天」
最高產日 = 月產量紀錄.index(max(月產量紀錄)) + 1
最低產日 = 月產量紀錄.index(min(月產量紀錄)) + 1

# 走訪整份紀錄，數出達標天數
日目標 = 500
達標天數 = 0
for 產量 in 月產量紀錄:
    if 產量 >= 日目標:
        達標天數 += 1

print("=" * 34)
print(f"  本月產量分析（共 {len(月產量紀錄)} 天）")
print("=" * 34)
print(f"  月總產量：{月總量:,} 件")
print(f"  日平均量：{月平均:.1f} 件")
print(f"  最高產日：第 {最高產日} 天（{max(月產量紀錄)} 件）")
print(f"  最低產日：第 {最低產日} 天（{min(月產量紀錄)} 件）")
print(f"  達標天數：{達標天數} 天 / {len(月產量紀錄)} 天（日目標 {日目標} 件）")
print("=" * 34)
```

## 第五章　基本資料處理

### 5-1　資料篩選與排序

```python
# 篩選（Filtering）：找出符合條件的資料

良率記錄 = [0.976, 0.981, 0.968, 0.943, 0.972, 0.985, 0.961, 0.978, 0.924, 0.969]

目標 = 0.95
未達標記錄 = []

for i, 良率 in enumerate(良率記錄):
    if 良率 < 目標:
        未達標記錄.append((i + 1, 良率))  # 儲存 (天數, 良率) 的配對

print("未達標日期：")
for 天數, 良率 in 未達標記錄:
    print(f"  第 {天數:2d} 天：{良率:.1%}")
print(f"未達標天數：{len(未達標記錄)} 天 / {len(良率記錄)} 天")
```

### 5-2　完整案例：感測器資料分析流程

## 第六章　效率與時間複雜度概念

> **給同學的提醒**：這一章的程式範例僅供參考，不會考。教材裡「對半翻最多 11 次」「泡沫排序約比較 n×n 次」這些數字，這裡用程式實際數一遍，證明它們是算得出來的。

### 6-1　為何需要效率分析：查字典的兩種找法

要在一本 2,000 頁的字典裡找一個字，有兩種找法：

- **一頁一頁翻**：從第 1 頁開始翻，運氣最差時要翻完整本
- **每次翻到中間**：翻過頭就往前找一半，還不夠就往後找一半，每翻一次就把還沒排除的範圍砍掉一半

對半找最多要翻幾次，可以先用算的：

$$2^{10} = 1024 < 2000 < 2048 = 2^{11}$$

所以最多 11 次。下面用程式實際數一遍，再把字典加厚到 4,000 頁比較看看。

```python
# ── 兩種找法在「運氣最差」時各要翻幾次 ──


def 逐頁翻(總頁數, 目標頁):
    """從第 1 頁開始一頁一頁翻，回傳翻了幾次"""
    次數 = 0
    for 頁 in range(1, 總頁數 + 1):
        次數 += 1
        if 頁 == 目標頁:
            break
    return 次數


def 對半翻(總頁數, 目標頁):
    """每次翻到剩餘範圍的中間，回傳翻了幾次"""
    次數 = 0
    最前, 最後 = 1, 總頁數
    while 最前 <= 最後:
        次數 += 1
        中間 = (最前 + 最後) // 2
        if 中間 == 目標頁:
            break
        elif 中間 < 目標頁:
            最前 = 中間 + 1  # 目標在後半，前半全部排除
        else:
            最後 = 中間 - 1  # 目標在前半，後半全部排除
    return 次數


for 總頁數 in [2000, 4000]:
    # 逐頁翻的最差情況：目標在最後一頁
    逐頁最差 = 逐頁翻(總頁數, 總頁數)
    # 對半翻的最差情況：把每一頁都當目標試一遍，取翻最多次的那一次
    對半最差 = max(對半翻(總頁數, 目標頁) for 目標頁 in range(1, 總頁數 + 1))
    print(f"字典 {總頁數:,} 頁：逐頁翻最多 {逐頁最差:,} 次，對半翻最多 {對半最差} 次")

print("\n頁數加倍 → 逐頁翻多翻 2,000 次，對半翻只多翻 1 次")
```

### 6-2　O(1)——常數時間

無論資料量 n 有多大，執行時間 **固定不變**。
用索引直接存取列表的某一格就是 O(1)：不必從頭找，直接跳到那一格。

```python
# ── 用索引存取：不管列表多長，都只做 1 個步驟 ──
for 資料量 in [5, 5_000, 500_000]:
    產量紀錄 = list(range(資料量))  # 建立一個有「資料量」個元素的列表
    步驟數 = 0

    第四個 = 產量紀錄[3]  # 直接跳到索引 3，不必經過前面的元素
    步驟數 += 1

    print(f"列表有 {資料量:>7,} 個元素 → 取出第 4 個元素用了 {步驟數} 個步驟")
```

### 6-3　O(n)——線性時間

執行時間 **與 n 成正比**：資料量變成 10 倍，步驟數也變成 10 倍。
線性搜尋從頭到尾逐一檢查，最差的情況是目標排在最後一筆，資料有幾筆就得比較幾次。

```python
# ── 線性搜尋：數一數最差情況要比較幾次 ──


def 線性搜尋(資料, 目標):
    """從頭到尾逐一比對，回傳（找到的索引, 比較次數）"""
    比較次數 = 0
    for 索引, 值 in enumerate(資料):
        比較次數 += 1
        if 值 == 目標:
            return 索引, 比較次數
    return None, 比較次數


for 資料量 in [100, 1_000, 10_000]:
    料號清單 = [f"MAT-{編號:05d}" for 編號 in range(1, 資料量 + 1)]
    目標料號 = 料號清單[-1]  # 最差情況：要找的料號排在最後一筆
    索引, 比較次數 = 線性搜尋(料號清單, 目標料號)
    print(f"在 {資料量:>6,} 筆料號中找 {目標料號} → 比較 {比較次數:>6,} 次")
```

### 6-4　O(n²)——平方時間

執行時間 **與 n 的平方成正比**：資料量變成 10 倍，步驟數變成約 100 倍。
泡沫排序每一輪把相鄰元素兩兩比較一遍（$n-1$ 次），這樣的輪次要跑 $n-1$ 輪，所以比較次數是：

$$(n-1) \times (n-1) \approx n^2$$

```python
# ── 泡沫排序：數一數總共比較幾次 ──


def 泡沫排序(資料):
    """由小到大排序，回傳（排序後的新列表, 比較次數）"""
    結果 = list(資料)  # 複製一份，不改動原本的列表
    筆數 = len(結果)
    比較次數 = 0
    for 輪 in range(筆數 - 1):  # 總共跑 n-1 輪
        for 位置 in range(筆數 - 1):  # 每一輪比較 n-1 組相鄰元素
            比較次數 += 1
            if 結果[位置] > 結果[位置 + 1]:
                結果[位置], 結果[位置 + 1] = 結果[位置 + 1], 結果[位置]
    return 結果, 比較次數


for 資料量 in [10, 100, 1_000]:
    待排序資料 = list(range(資料量, 0, -1))  # 由大到小的資料，要排成由小到大
    排序結果, 比較次數 = 泡沫排序(待排序資料)
    排序正確 = 排序結果 == sorted(待排序資料)
    print(f"排序 {資料量:>5,} 筆 → 比較 {比較次數:>7,} 次（n×n = {資料量 * 資料量:>9,}），排序正確：{排序正確}")
```

### 6-5　三種複雜度放在一起比

假設每個基本步驟花 1 微秒（μs，百萬分之一秒），算出三種複雜度在不同資料量下各要多久。
以資料量 100 萬筆的 O(n²) 為例：

$$10^{6} \times 10^{6} = 10^{12}\ \mu s = 10^{6}\ 秒 \approx 11.6\ 天$$

```python
# ── 每個步驟 1 微秒，三種複雜度各要多久 ──


def 換算時間(微秒):
    """把微秒換成容易讀的單位"""
    秒 = 微秒 / 1_000_000
    if 秒 >= 86_400:  # 一天有 86,400 秒
        return f"{秒 / 86_400:.1f} 天"
    if 秒 >= 1:
        return f"{秒:g} 秒"
    if 微秒 >= 1_000:
        return f"{微秒 / 1_000:g} ms"
    return f"{微秒:g} μs"


print(f"{'n（資料量）':<10}{'O(1)':>10}{'O(n)':>10}{'O(n²)':>10}")
print("-" * 44)
for 資料量 in [10, 1_000, 1_000_000]:
    常數時間 = 換算時間(1)
    線性時間 = 換算時間(資料量)
    平方時間 = 換算時間(資料量 * 資料量)
    print(f"{資料量:<15,}{常數時間:>10}{線性時間:>10}{平方時間:>10}")
```

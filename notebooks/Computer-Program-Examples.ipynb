## 第一篇：程式設計與運算思維

## 第一章　問題解決與運算思維

運算思維的四個核心面向：**問題分解、模式辨識、抽象化、演算法思考**。  
本章用一個「生產線良率分析」問題，示範如何用這四個思維拆解並解決問題。

### 情境：工廠有三條生產線，需要分析本週每條線的良率，並找出低於目標的生產線。

### 步驟 1　問題分解（Decomposition）

把「分析三條生產線的良率」這個大問題，拆解成幾個小任務：
1. 計算每條線的良率
2. 判斷每條線是否達標（目標：良率 ≥ 95%）
3. 列出未達標的生產線
4. 計算全廠平均良率

每個小任務都可以獨立撰寫，最後組合起來就是完整的分析。

```python
# ── 步驟 1：準備資料（每條生產線的總產量與不良品數）──

# 生產線名稱
lines = ["A線", "B線", "C線"]

# 各線本週總產量（件）
total = [500, 320, 410]

# 各線本週不良品數（件）
defects = [12, 8, 31]

print("資料準備完成")
print(f"生產線：{lines}")
print(f"總產量：{total}")
print(f"不良品：{defects}")
```

### 步驟 2　模式辨識（Pattern Recognition）

觀察三條線的計算方式完全相同：`良率 = (總產量 - 不良品數) / 總產量 × 100`  
這個重複的計算模式，可以用**迴圈**來統一處理。

```python
# ── 步驟 2：辨識出重複模式 → 用迴圈計算每條線的良率 ──

yield_rates = []  # 用來儲存每條線的良率

for i in range(3):  # i 依序為 0, 1, 2
    good = total[i] - defects[i]  # 良品數
    rate = good / total[i] * 100  # 良率（%）
    yield_rates.append(rate)  # 加入清單
    print(f"{lines[i]}：良品 {good} 件，良率 {rate:.1f}%")
```

### 步驟 3　抽象化（Abstraction）

「計算良率」這個動作可以包裝成一個**函數**，  
呼叫時只需給「總產量」和「不良品數」，不需在意內部細節。  
這就是抽象化——隱藏實作細節，暴露清楚的介面。

```python
# ── 步驟 3：將重複邏輯抽象化為函數 ──


def 計算良率(總產量, 不良品數):
    """輸入總產量與不良品數，回傳良率百分比"""
    良品數 = 總產量 - 不良品數
    良率 = 良品數 / 總產量 * 100
    return 良率


# 函數抽象化後，呼叫非常直觀
print(f"A線良率：{計算良率(500, 12):.1f}%")
print(f"B線良率：{計算良率(320,  8):.1f}%")
print(f"C線良率：{計算良率(410, 31):.1f}%")
```

### 步驟 4　演算法思考（Algorithmic Thinking）

將整個分析流程，寫成**清楚的步驟序列**：
1. 計算每條線的良率
2. 與目標值比較，判斷是否達標
3. 列出未達標的生產線
4. 計算並顯示全廠平均良率

```python
# ── 步驟 4：完整演算法——從資料到結論 ──

目標良率 = 95.0  # 目標：良率需達 95%
未達標清單 = []

print("=" * 35)
print("本週生產線良率分析報告")
print("=" * 35)

# 步驟 1 & 2：計算良率並判斷達標
for i in range(3):
    rate = 計算良率(total[i], defects[i])
    達標 = rate >= 目標良率
    狀態 = "✓ 達標" if 達標 else "✗ 未達標"
    print(f"{lines[i]}：{rate:.1f}%  {狀態}")
    if not 達標:
        未達標清單.append(lines[i])

# 步驟 3：列出未達標生產線
print("-" * 35)
if 未達標清單:
    print(f"需要改善的生產線：{未達標清單}")
else:
    print("所有生產線均達標！")

# 步驟 4：全廠平均良率
全廠良率 = sum(yield_rates) / len(yield_rates)
print(f"全廠平均良率：{全廠良率:.1f}%")
print("=" * 35)
```

### 3-2　for 迴圈

```python
# 用 for 迴圈處理一週的生產紀錄

星期 = ["週一", "週二", "週三", "週四", "週五"]
每日產量 = [480, 512, 498, 467, 521]
每日不良 = [12, 8, 15, 22, 9]

目標良率 = 95.0
總產量 = 0
低於目標天數 = 0

print(f"{'日期':4}  {'產量':>6}  {'不良':>4}  {'良率':>7}  {'狀態'}")
print("-" * 40)

for i in range(5):  # i 從 0 到 4
    良率 = (每日產量[i] - 每日不良[i]) / 每日產量[i] * 100
    總產量 += 每日產量[i]  # 累計總產量
    狀態 = "✓" if 良率 >= 目標良率 else "✗"
    if 良率 < 目標良率:
        低於目標天數 += 1
    print(f"{星期[i]:4}  {每日產量[i]:>6}  {每日不良[i]:>4}  {良率:>6.1f}%  {狀態}")

print("-" * 40)
print(f"週總產量：{總產量} 件")
print(f"低於目標天數：{低於目標天數} 天")
```

### 3-3　while 迴圈

```python
# 模擬生產線運作：生產到達配額為止

配額 = 100  # 今日生產目標
已生產 = 0
批次大小 = 12  # 每次加工一個批次
批次編號 = 0

print(f"今日配額：{配額} 件")
print("-" * 30)

while 已生產 < 配額:
    批次編號 += 1
    # 避免超產：本批次只生產到剛好達標
    本批產量 = min(批次大小, 配額 - 已生產)
    已生產 += 本批產量
    print(f"批次 {批次編號:02d}：生產 {本批產量} 件，累計 {已生產} 件")

print("-" * 30)
print(f"配額完成！共執行 {批次編號} 個批次")
```

### 3-4　break 與 continue

```python
# break：偵測到嚴重異常立即停止

溫度資料 = [22.1, 23.4, 21.8, None, 22.9, 31.5, 23.1]
#                                ↑            ↑
#                            感測器離線    緊急高溫

緊急上限 = 30.0

print("開始逐筆分析感測器資料...")

for i, 溫度 in enumerate(溫度資料):
    # continue：跳過空值（感測器離線）
    if 溫度 is None:
        print(f"  第 {i+1} 筆：感測器離線，略過")
        continue

    # break：溫度超過緊急上限，立即中止
    if 溫度 > 緊急上限:
        print(f"  第 {i+1} 筆：{溫度}°C ── 🔴 緊急高溫！停止分析，執行停機程序")
        break

    print(f"  第 {i+1} 筆：{溫度}°C ── 正常")

print("分析結束")
```

## 第四章　函數與模組化

### 4-1　定義與呼叫函數

```python
# 函數定義：計算良率
def 計算良率(總產量, 不良品數):
    """輸入總產量與不良品數，回傳良率百分比（0~100）"""
    if 總產量 == 0:
        return 0  # 避免除以零
    良品數 = 總產量 - 不良品數
    return 良品數 / 總產量 * 100


# 函數呼叫：同一個函數可以重複使用
A線良率 = 計算良率(500, 12)
B線良率 = 計算良率(320, 8)
C線良率 = 計算良率(410, 31)

print(f"A線良率：{A線良率:.1f}%")
print(f"B線良率：{B線良率:.1f}%")
print(f"C線良率：{C線良率:.1f}%")
```

```python
# 函數可以有預設參數值
def 判斷達標(良率, 目標=95.0):
    """判斷良率是否達到目標，目標預設為 95%"""
    if 良率 >= 目標:
        return True, "達標 ✓"
    else:
        差距 = 目標 - 良率
        return False, f"未達標 ✗（差 {差距:.1f}%）"


# 使用預設目標 95%
達標, 訊息 = 判斷達標(A線良率)
print(f"A線：{訊息}")

達標, 訊息 = 判斷達標(C線良率)
print(f"C線：{訊息}")

# 指定不同目標
達標, 訊息 = 判斷達標(C線良率, 目標=90.0)
print(f"C線（目標90%）：{訊息}")
```

### 4-2　模組化設計——將生產報表系統拆成多個函數

每個函數只負責一件事，最後在主程式中組合使用。

```python
# ── 模組1：計算單條生產線的統計數字 ──
def 分析生產線(名稱, 總產量, 不良品數, 目標良率=95.0):
    """回傳一條生產線的分析結果（字典格式）"""
    良率 = 計算良率(總產量, 不良品數)
    達標, 狀態 = 判斷達標(良率, 目標良率)
    return {"名稱": 名稱, "總產量": 總產量, "良率": 良率, "達標": 達標, "狀態": 狀態}


# ── 模組2：列印單條生產線的資訊 ──
def 列印生產線結果(結果):
    print(
        f"  {結果['名稱']}：產量 {結果['總產量']} 件，"
        f"良率 {結果['良率']:.1f}%  {結果['狀態']}"
    )


# ── 模組3：產生完整報表 ──
def 產生日報表(生產資料):
    """輸入多條生產線資料，輸出完整日報"""
    print("=" * 45)
    print("           生產日報表")
    print("=" * 45)

    所有良率 = []
    未達標線 = []

    for 線名, 產量, 不良 in 生產資料:
        結果 = 分析生產線(線名, 產量, 不良)
        列印生產線結果(結果)
        所有良率.append(結果["良率"])
        if not 結果["達標"]:
            未達標線.append(結果["名稱"])

    print("-" * 45)
    平均良率 = sum(所有良率) / len(所有良率)
    print(f"  全廠平均良率：{平均良率:.1f}%")
    if 未達標線:
        print(f"  ⚠ 需要改善：{未達標線}")
    else:
        print("  ✓ 所有生產線均達標")
    print("=" * 45)


# ── 主程式：呼叫報表函數 ──
今日資料 = [
    ("A線", 500, 12),
    ("B線", 320, 8),
    ("C線", 410, 31),
    ("D線", 280, 4),
]

產生日報表(今日資料)
```

### 5-2　資料篩選與排序

```python
# 篩選（Filtering）：找出符合條件的資料

良率記錄 = [0.976, 0.981, 0.968, 0.943, 0.972, 0.985, 0.961, 0.978, 0.924, 0.969]

目標 = 0.95
未達標記錄 = []

for i, 良率 in enumerate(良率記錄):
    if 良率 < 目標:
        未達標記錄.append((i + 1, 良率))  # 儲存 (天數, 良率) 的配對

print("未達標日期：")
for 天數, 良率 in 未達標記錄:
    print(f"  第 {天數:2d} 天：{良率:.1%}")
print(f"未達標天數：{len(未達標記錄)} 天 / {len(良率記錄)} 天")
```

### 5-3　完整案例：感測器資料分析流程

```python
# 完整示範：讀取 → 清洗 → 分析 → 輸出

# ── 模擬讀取一天的溫度感測器資料（每小時一筆，共24筆）──
原始溫度資料 = [
    22.1,
    22.3,
    22.0,
    21.8,
    21.5,
    21.9,  # 0:00 ~ 5:00
    22.4,
    23.1,
    23.5,
    23.8,
    24.2,
    24.5,  # 6:00 ~ 11:00
    None,
    24.8,
    26.1,
    25.9,
    24.3,
    23.7,  # 12:00（離線）~ 17:00
    23.2,
    22.8,
    22.5,
    22.1,
    21.9,
    22.0,  # 18:00 ~ 23:00
]

上限 = 25.0
下限 = 20.0

# ── 步驟1：清洗資料（移除空值）──
有效資料 = [t for t in 原始溫度資料 if t is not None]
缺失筆數 = len(原始溫度資料) - len(有效資料)

# ── 步驟2：彙整統計 ──
平均溫度 = sum(有效資料) / len(有效資料)
最高溫度 = max(有效資料)
最低溫度 = min(有效資料)

# ── 步驟3：找出異常時段 ──
異常紀錄 = []
for i, 溫度 in enumerate(原始溫度資料):
    if 溫度 is None:
        continue
    if 溫度 > 上限 or 溫度 < 下限:
        異常紀錄.append((i, 溫度))

# ── 步驟4：輸出報告 ──
print("=" * 40)
print("  溫度感測器日分析報告")
print("=" * 40)
print(f"  資料筆數：{len(原始溫度資料)} 筆（缺失 {缺失筆數} 筆）")
print(f"  平均溫度：{平均溫度:.1f}°C")
print(f"  最高溫度：{最高溫度:.1f}°C")
print(f"  最低溫度：{最低溫度:.1f}°C")
print(f"  正常範圍：{下限}°C ~ {上限}°C")
print("-" * 40)
if 異常紀錄:
    print(f"  ⚠ 異常時段（共 {len(異常紀錄)} 筆）：")
    for 時間, 溫度 in 異常紀錄:
        print(f"     {時間:02d}:00  {溫度:.1f}°C  ← 超出範圍")
else:
    print("  ✓ 全天溫度均在正常範圍內")
print("=" * 40)
```

### 7-2　存取與走訪

**隨機存取**是「直接跳到第 n 格」，**走訪**是「從頭到尾跑一遍」。
兩者是列表最常用的操作，也是後面所有資料處理的基礎。

```python
每日產量 = [480, 512, 498, 467, 521, 489, 503]

# ── 隨機存取（Random Access）：直接跳到指定索引 ──
# 不管列表多長，取出任何一個元素的時間都一樣快
print("【隨機存取】")
print(f"第 3 天產量：{每日產量[2]} 件")
print(f"第 7 天產量：{每日產量[6]} 件")

# ── 走訪（Traversal）：從頭到尾，一個都不漏 ──
print("\n【走訪】")
累計 = 0
for 第幾天, 產量 in enumerate(每日產量, start=1):
    累計 += 產量
    print(f"第 {第幾天} 天：{產量:>4} 件，累計 {累計:>5} 件")
```

## 第八章　堆疊與佇列（Stack & Queue）

### 8-1　堆疊（Stack）——後進先出（LIFO）

像一疊盤子：只能從最上面放（Push）、從最上面拿（Pop），不能從中間抽。

```python
# ── 堆疊（Stack）：用 Python 的 list 就能實作 ──
# append() 就是 Push（推入頂端）、pop() 就是 Pop（彈出頂端）
操作紀錄 = []

操作紀錄.append("修改工單 W-1001 數量")
操作紀錄.append("新增工單 W-1002")
操作紀錄.append("刪除工單 W-0998")

print(f"目前堆疊（最左邊是底部）：{操作紀錄}")
print(f"堆疊頂端（最後放進去的）：{操作紀錄[-1]}")

# 使用者連按三次「復原」——每次都從頂端彈出
print("\n【連按三次復原】")
while 操作紀錄:
    最後操作 = 操作紀錄.pop()
    print(f"  復原：{最後操作}（堆疊剩 {len(操作紀錄)} 筆）")
```

**工業應用：設備的啟動與關閉順序**

有些設備的關機必須以啟動的相反順序進行（後啟動的先關），才不會出安全問題。
這個「反過來」的需求，用堆疊表達最自然——完全不必自己去記順序。

```python
# ── 工業應用：設備開機依序啟動，關機必須反過來拆 ──
啟動順序 = ["主電源", "冷卻系統", "液壓系統", "主軸馬達", "自動送料"]

安全堆疊 = []
print("【開機】依序啟動，每啟動一項就推入堆疊")
for 步驟, 設備 in enumerate(啟動順序, start=1):
    安全堆疊.append(設備)
    print(f"  步驟 {步驟}：啟動 {設備}")

print("\n【關機】從堆疊頂端逐一彈出，順序自動反過來")
步驟 = 1
while 安全堆疊:
    設備 = 安全堆疊.pop()
    print(f"  步驟 {步驟}：關閉 {設備}")
    步驟 += 1
```

### 8-2　佇列（Queue）——先進先出（FIFO）

像排隊結帳：先到的先被服務。工單排程、訂單處理、列印排隊都是這個模型。

```python
from collections import deque

# ── 佇列（Queue）：deque 是 Python 標準函式庫提供的雙向佇列 ──
# append() 就是 Enqueue（加入尾端）、popleft() 就是 Dequeue（取出頭部）
工單佇列 = deque()

for 工單 in ["W-1001", "W-1002", "W-1003"]:
    工單佇列.append(工單)
    print(f"工單 {工單} 下達 → 目前排隊：{list(工單佇列)}")

# 機台一次只能做一張，做完才輪到下一張
加工時間 = {"W-1001": 45, "W-1002": 30, "W-1003": 60}
目前時刻 = 0

print("\n【機台開始加工】")
while 工單佇列:
    工單 = 工單佇列.popleft()
    開始 = 目前時刻
    目前時刻 += 加工時間[工單]
    print(f"  第 {開始:>3} ~ {目前時刻:>3} 分：加工 {工單}（還有 {len(工單佇列)} 張在等）")
```

### 8-3　同一批資料，兩種取出順序

```python
# ── 同一批零件，用堆疊和用佇列取出的順序完全不同 ──
進站順序 = ["A", "B", "C", "D"]

堆疊 = []
佇列 = deque()
for 零件 in 進站順序:
    堆疊.append(零件)
    佇列.append(零件)

堆疊取出 = []
while 堆疊:
    堆疊取出.append(堆疊.pop())      # 從頂端拿 → 後進先出

佇列取出 = []
while 佇列:
    佇列取出.append(佇列.popleft())  # 從頭部拿 → 先進先出

print(f"零件進站順序　　　：{' → '.join(進站順序)}")
print(f"堆疊取出（LIFO）：{' → '.join(堆疊取出)}")
print(f"佇列取出（FIFO）：{' → '.join(佇列取出)}")
```

## 第九章　集合與映射（Set / Map）

### 9-1　集合（Set）——只留下不重複的值

```python
# ── 集合（Set）：丟進去的重複值會自動被吃掉 ──
本月不良紀錄 = ["刮傷", "尺寸偏差", "刮傷", "毛邊", "尺寸偏差",
                "刮傷", "破損", "毛邊", "刮傷", "尺寸偏差"]

不良類型 = set(本月不良紀錄)

print(f"原始紀錄共 {len(本月不良紀錄)} 筆")
print(f"不重複的不良類型共 {len(不良類型)} 種：{sorted(不良類型)}")

# 再加入一個已經有的值，集合不會變大
不良類型.add("刮傷")
print(f"再加一次「刮傷」後，仍然是 {len(不良類型)} 種")

# 集合最常用的操作：快速判斷「有沒有出現過」
for 待查 in ["破損", "氧化"]:
    結果 = "有出現" if 待查 in 不良類型 else "沒出現"
    print(f"  本月是否出現「{待查}」？{結果}")
```

**集合運算**：集合可以直接做交集、聯集、差集，用來比對兩批資料的異同特別方便。

```python
# ── 集合運算：兩個月的不良類型比一比 ──
標準不良類型 = {"刮傷", "尺寸偏差", "毛邊", "破損", "氧化", "變形"}
本月出現 = {"刮傷", "尺寸偏差", "毛邊", "破損"}
上月出現 = {"刮傷", "氧化", "毛邊"}

print(f"本月都沒出現的類型（差集 -）：{sorted(標準不良類型 - 本月出現)}")
print(f"兩個月都有的老問題（交集 &）：{sorted(本月出現 & 上月出現)}")
print(f"本月新冒出來的問題（差集 -）：{sorted(本月出現 - 上月出現)}")
print(f"兩個月出現過的全部（聯集 |）：{sorted(本月出現 | 上月出現)}")
```

### 9-2　映射（Map）——用鍵直接查到值

Python 的字典（`dict`）就是映射：一個 **鍵（Key）** 對應一個 **值（Value）**，
不論主檔有多少筆資料，查詢時間幾乎不變。

```python
# ── 映射（Map / Python 的 dict）：用鍵（料號）直接查到值（物料資料）──
物料主檔 = {
    "MAT-001": {"名稱": "齒輪", "單位": "個", "安全庫存": 100},
    "MAT-002": {"名稱": "軸承", "單位": "個", "安全庫存": 250},
    "MAT-003": {"名稱": "鏈條", "單位": "條", "安全庫存": 60},
}

料號 = "MAT-002"
資料 = 物料主檔[料號]
print(f"查詢 {料號} → {資料['名稱']}（{資料['單位']}），安全庫存 {資料['安全庫存']}")

# 查不存在的鍵會讓程式中斷，用 get() 給一個預設值比較安全
print(f"查詢 MAT-999 → {物料主檔.get('MAT-999', '查無此料號')}")

# 走訪整本主檔
print("\n【物料主檔一覽】")
for 料號, 資料 in 物料主檔.items():
    print(f"  {料號}　{資料['名稱']}　安全庫存 {資料['安全庫存']:>4} {資料['單位']}")
```

### 9-3　映射的延伸應用：次數統計與查詢效率

```python
# ── 映射的另一個常見用途：統計次數 ──
# 沿用第 9-1 節的「本月不良紀錄」
不良次數 = {}
for 類型 in 本月不良紀錄:
    不良次數[類型] = 不良次數.get(類型, 0) + 1  # 沒查到就從 0 開始算

# 依次數由多到少排序，找出最該優先改善的項目
排序結果 = sorted(不良次數.items(), key=lambda 項: 項[1], reverse=True)

print("【本月不良統計（由多到少）】")
for 類型, 次數 in 排序結果:
    print(f"  {類型}：{次數} 次　{'█' * 次數}")

print(f"\n最該優先改善：{排序結果[0][0]}（占 {排序結果[0][1] / len(本月不良紀錄):.0%}）")
```

```python
# ── 為什麼要用映射？和「列表逐一搜尋」比一比 ──
物料列表 = [("MAT-001", "齒輪"), ("MAT-002", "軸承"), ("MAT-003", "鏈條")]
物料字典 = {"MAT-001": "齒輪", "MAT-002": "軸承", "MAT-003": "鏈條"}

def 用列表查(目標料號):
    """從頭到尾一筆一筆比對，回傳結果與比對次數"""
    比對次數 = 0
    for 編號, 名稱 in 物料列表:
        比對次數 += 1
        if 編號 == 目標料號:
            return 名稱, 比對次數
    return "查無此料號", 比對次數

名稱, 次數 = 用列表查("MAT-003")
print(f"列表逐一搜尋 MAT-003 → {名稱}，比對了 {次數} 次")
print(f"映射直接查詢 MAT-003 → {物料字典['MAT-003']}，比對 1 次")
print("\n主檔只有 3 筆時感覺不出差別；")
print("若主檔有 10,000 筆，列表平均要比對 5,000 次，映射仍然只要 1 次。")
```
